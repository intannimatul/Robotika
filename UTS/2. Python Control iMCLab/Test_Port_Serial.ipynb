```python
import serial

arduinoData = serial.Serial('COM4', 115200, timeout=1)
print("Terhubung ke ESP32 di COM4!")

try:
    while True:
        myData = input('Kirimkan perintah (1=ON, 0=OFF, q=Keluar): ').strip()
        if myData == "1":
            arduinoData.write(b'1\n')
            print(">> Motor ON")
        elif myData == "0":
            arduinoData.write(b'0\n')
            print(">> Motor OFF")
        elif myData.lower() == "q":
            print("Menghentikan pengujian...")
            break
finally:
    arduinoData.close()
    print("Port COM4 berhasil ditutup aman.")
```
